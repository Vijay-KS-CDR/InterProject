# Agricultural AI/ML Project: Soil Nutrient Estimation
## Task 1: Dataset Loading and Exploratory Data Inspection

This notebook performs basic inspection of the agricultural remote-sensing dataset `synthetic_crop_data.csv`.
The goal of the overall project is to estimate soil nutrient status (**Nitrogen (N), Phosphorus (P), Potassium (K)**) using crop canopy features and remote sensing indices.

### Step 1: Import Libraries and Load the Dataset
We use `pandas` for tabular data manipulation, and `matplotlib` & `seaborn` for visualizations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configure visual theme
sns.set_theme(style="whitegrid")
plt.rcParams.update({"font.sans-serif": "DejaVu Sans", "font.size": 10})

# 1. Load the CSV dataset
df = pd.read_csv("synthetic_crop_data.csv")
print("Dataset loaded successfully!")
```

### Step 2: Display the First 5 Rows
`df.head()` allows us to preview the top rows and inspect the structure of each column.

```python
df.head()
```

### Step 3 & 4: Dimensions and Column Names
- `df.shape` returns (number of rows, number of columns).
- `df.dtypes` lists the datatype for each column.

```python
print(f"Total Rows: {df.shape[0]}")
print(f"Total Columns: {df.shape[1]}\n")
print("Column Names and Data Types:")
print(df.dtypes)
```

### Step 5: Check for Missing Values
`df.isnull().sum()` checks how many empty (NaN) cells exist in each column.

```python
missing_counts = df.isnull().sum()
print("Missing Values per Column:")
print(missing_counts)
print("\nTotal Missing Values in Dataset:", missing_counts.sum())
```

### Step 6: Summary Statistics with `describe()`
`describe()` calculates count, mean, standard deviation, min, 25%, 50% (median), 75%, and max.

```python
df.describe().round(3)
```

### Step 7 & 8: Inspect Categorical Columns (`Crop` and `Growth Stage`)
We count the occurrences of each crop and growth stage to check class balance.

```python
print("--- Crop Distribution ---")
print(df["Crop"].value_counts())

print("\n--- Growth Stage Distribution ---")
print(df["Growth Stage"].value_counts())
```

### Step 9: Check for Duplicate Rows
`df.duplicated().sum()` identifies if any rows are exact duplicates of others.

```python
num_duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {num_duplicates}")
```

### Step 10: Minimum & Maximum Values of Key Features
Key remote-sensing indices (NDVI, EVI, FVC, LAI, Biomass) and target nutrients (N, P, K).

```python
key_features = ["NDVI", "EVI", "FVC", "LAI", "Biomass", "N", "P", "K"]
min_max_summary = pd.DataFrame({
    "Feature": key_features,
    "Minimum": [df[col].min() for col in key_features],
    "Maximum": [df[col].max() for col in key_features],
    "Range": [df[col].max() - df[col].min() for col in key_features]
})
min_max_summary
```

### Step 11: Histograms of Numerical Features
Visualizing distributions helps us see the spread, central tendency, and potential outliers.

```python
numerical_cols = ["NDVI", "EVI", "FVC", "LAI", "NDVI Texture", "Biomass", "N", "P", "K"]
fig, axes = plt.subplots(3, 3, figsize=(14, 10))
axes = axes.flatten()

for idx, col in enumerate(numerical_cols):
    sns.histplot(df[col], kde=True, ax=axes[idx], color="forestgreen", edgecolor="black", alpha=0.6)
    axes[idx].set_title(f"Distribution of {col}", fontweight="bold")

plt.tight_layout()
plt.show()
```

### Step 12: Correlation Matrix and Heatmap
A correlation heatmap shows linear relationships between remote-sensing indicators and soil nutrient values (N, P, K).

```python
corr_matrix = df[numerical_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, square=True, linewidths=0.5)
plt.title("Correlation Heatmap: Remote Sensing & Soil Nutrients (N, P, K)", fontweight="bold", pad=12)
plt.tight_layout()
plt.show()
```
